# Garmin daily stats
Steps, sleep, resting HR, stress, and body battery pulled from Garmin Connect.

```python
import pandas as pd
import plotly.express as px

# load garmin data, oldest day first
df = pd.read_csv("../data/garmin.csv", parse_dates=["date"]).sort_values("date")

# quick summary: averages, min, max for every column
df.describe().round(1)
```

### Daily trends
Each metric over time, with a 7-day rolling average to smooth out day-to-day noise.

```python
# metrics to chart
metrics = ["sleep_hours", "resting_hr", "stress_avg", "body_battery_wake", "steps"]

# 7-day rolling average for each metric
weekly = df.set_index("date")[metrics].rolling(7, min_periods=3).mean().reset_index()

# reshape so plotly can make one mini chart per metric
long = weekly.melt(id_vars="date", var_name="metric", value_name="value")

fig = px.line(
    long, x="date", y="value",
    facet_row="metric", height=900,
    template="plotly_white",
    title="Garmin trends (7-day average)",
)
fig.update_yaxes(matches=None, title="")
fig.update_xaxes(title="")
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[1]))
fig.update_traces(line_color="#ff6b6b")
fig.show()
```
